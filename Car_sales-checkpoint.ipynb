```python
df_std = pd.DataFrame()
for col in df_num.columns:
    m = df_num[col].mean()
    s = df_num[col].std()
    df_std[col + '_n'] = (df_num[col] - m) / s
df_std.head(10)
```

```python
fig = plt.figure(figsize = (20, 8))
ax = fig.add_axes([0, 0, 1, 1])
ax.set_xticklabels(df_std.columns)

data = []
for col in df_std.columns:
    data.append(df_std[col])

bp = ax.boxplot(data, widths=0.5)
plt.show()
```

```python
x = df['Width']
y= df['Length']
plt.scatter(x, y)
plt.xlabel('Width')
plt.ylabel('Lenght')
plt.show()
```

```python
mx, sx = x.mean(), x.std()
my, sy = y.mean(), y.std()
Xs = (x - mx) / sx
Ys = (y - my) / sy
plt.scatter(Xs, Ys)
plt.xlabel('width_s')
plt.ylabel('Lenght_s')
plt.show()
```

```python
clustering = DBSCAN(eps=0.3, min_samples=3).fit(list(zip(Xs, Ys)))
labels = clustering.labels_
print(labels)
```

```python
def random_color():
    return (random.randint(0, 255)/255, random.randint(0, 255)/255, random.randint(0, 255)/255)

palette = {}
colors = []


for key in set(labels):
    if key == -1:
        palette[key] = (0, 0, 0)
    else:
        palette[key] = random_color()

for lb in labels:
    colors.append(palette[lb])  

plt.scatter(Xs, Ys, c=colors)
plt.xlabel('Width')
plt.ylabel('Length')           
plt.show()
```

```python
X = df['Engine_size']
Y = df['Horsepower']
```

```python
print('Info')
print('X:', *X, sep='\t')
print('Y:', *Y, sep='\t')
Xnew = X[Y < Y.max()]
Ynew = Y[Y < Y.max()]
print('Clear INFO')
print('X:', Xnew, sep='\t')
print('Y:', Ynew, sep='\t')
```

```python
m0 = sps.linregress(X, Y)
m1 = sps.linregress(Xnew, Ynew)
```

```python
x0 = np.array([X.min(), X.max()])
y0 = m0.intercept + m0.slope * x0
x1= np.array([Xnew.min(), Xnew.max()])
y1 = m1.intercept + m1.slope * x1
plt.figure(figsize=(5, 10))
plt.scatter(X, Y)
plt.plot(x0, y0, color='red', label=f'Y={round(m0.intercept, 3)} +{round(m0.slope, 3)} x')
plt.plot(x1, y1, color='green', label=f'Y={round(m1.intercept, 3)}+{round(m1.slope, 3)} x')
plt.xlabel('X')
plt.ylabel('Y')
plt.legend()
plt.grid()
plt.show()
```
